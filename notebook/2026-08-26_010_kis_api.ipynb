```python
def balance():
    r = kis_api.domestic_stock_inquire_balance()
    pos = dict()
    for item in r.get('output1'):
        sym = item.get('pdno')
        size = item.get('hldg_qty')
        avg = item.get('pchs_avg_pric')
        pos[sym] = (size, avg)
    cash = r.get('output2')[0].get('dnca_tot_amt')
    equity = r.get('output2')[0].get('tot_evlu_amt')
    return pos, cash, equity
```
