```python
# ── 체결가 데이터 파싱 함수 ──
def parse_execution_data(data_str):
    """
    체결가 수신 데이터를 파싱하는 함수
    데이터는 "^"로 구분된 문자열
    """
    fields = data_str.split("^")        # "^" 구분자로 분리
    
    result = {
        "종목코드":    fields[0],         # 종목코드
        "체결시간":    fields[1],         # 체결시간 (HHMMSS)
        "현재가":      fields[2],         # 현재가
        "전일대비부호": fields[3],        # 1:상한 2:상승 3:보합 4:하한 5:하락
        "전일대비":    fields[4],         # 전일 대비 가격
        "등락률":      fields[5],         # 등락률
        "가중평균가":  fields[6],          # 가중 평균가
        "시가":        fields[7],
        "고가":        fields[8],
        "저가":        fields[9],
        "매도호가1":   fields[10],
        "매수호가1":   fields[11],
        "체결량":      fields[12],        # 이번 체결 수량
        "누적거래량":  fields[13],        # 오늘 총 거래량
        "누적거래대금": fields[14],       # 오늘 총 거래대금
    }
    return result

# ── 호가 데이터 파싱 함수 ──
def parse_orderbook_data(data_str):
    """
    호가 수신 데이터를 파싱하는 함수
    매도/매수 각 10단계 호가 포함
    """
    fields = data_str.split("^")
    
    result = {"종목코드": fields[0], "시간": fields[1]}
    
    # 매도호가 10단계 (높은가격 → 낮은가격)
    for i in range(10):
        result[f"매도호가{i+1}"] = fields[3 + i]      # 매도 가격
        result[f"매도잔량{i+1}"] = fields[23 + i]      # 매도 잔량
    
    # 매수호가 10단계 (높은가격 → 낮은가격)
    for i in range(10):
        result[f"매수호가{i+1}"] = fields[13 + i]      # 매수 가격
        result[f"매수잔량{i+1}"] = fields[33 + i]      # 매수 잔량
    
    return result
```

```python
def get_hashkey(data):
    """주문 데이터로부터 보안용 해시키를 생성하는 함수"""
    PATH = "uapi/hashkey"                       # 해시키 발급 경로
    URL = f"{URL_BASE}/{PATH}"
    headers = {
        "content-type": "application/json",
        "appkey": APP_KEY,                       # 앱 키
        "appsecret": APP_SECRET                  # 앱 시크릿
    }
    res = requests.post(URL, headers=headers, data=json.dumps(data))  # POST 요청
    hashkey = res.json()['HASH']                 # 응답에서 해시값 추출
    return hashkey

# ── 매수 주문 함수 ──
def buy_stock(code, qty="1", price="0"):
    """
    종목코드, 수량, 가격을 받아 매수 주문을 실행하는 함수
    price="0"이면 시장가 주문
    """
    PATH = "uapi/domestic-stock/v1/trading/order-cash"  # 주문 경로
    URL = f"{URL_BASE}/{PATH}"

    # 주문 데이터 구성
    data = {
        "CANO": CANO,                    # 계좌번호 앞 8자리
        "ACNT_PRDT_CD": ACNT_PRDT_CD,    # 계좌번호 뒤 2자리
        "PDNO": code,                     # 종목코드
        "ORD_DVSN": "01" if price == "0" else "00",  # 01:시장가, 00:지정가
        "ORD_QTY": qty,                   # 주문 수량
        "ORD_UNPR": price                 # 주문 단가 (시장가면 "0")
    }

    headers = {
        "Content-Type": "application/json",
        "authorization": f"Bearer {ACCESS_TOKEN}",
        "appKey": APP_KEY,
        "appSecret": APP_SECRET,
        "tr_id": "TTTC0802U",             # 실전 매수 (모의: VTTC0802U)
        "custtype": "P",                   # 개인
        "hashkey": get_hashkey(data)       # 보안용 해시키
    }

    res = requests.post(URL, headers=headers, data=json.dumps(data))  # POST 요청
    result = res.json()

    if result['rt_cd'] == '0':             # 성공
        print(f"매수 주문 성공! 주문번호: {result['output']['ODNO']}")
    else:
        print(f"매수 주문 실패: {result['msg1']}")
    return result

# ── 실행: 삼성전자 1주 시장가 매수 (모의투자에서 테스트!) ──
# buy_stock("005930", "1", "0")  # 주석 해제 후 실행

# ── 매도 주문 함수 ──
def sell_stock(code, qty="1", price="0"):
    """매도 주문 - 매수와 거의 동일하되 tr_id만 다름"""
    PATH = "uapi/domestic-stock/v1/trading/order-cash"
    URL = f"{URL_BASE}/{PATH}"

    data = {
        "CANO": CANO,
        "ACNT_PRDT_CD": ACNT_PRDT_CD,
        "PDNO": code,
        "ORD_DVSN": "01" if price == "0" else "00", # 01:시장가, 00:지정가
        "ORD_QTY": qty,
        "ORD_UNPR": price
    }

    headers = {
        "Content-Type": "application/json",
        "authorization": f"Bearer {ACCESS_TOKEN}",
        "appKey": APP_KEY,
        "appSecret": APP_SECRET,
        "tr_id": "TTTC0801U",             # 실전 매도 (모의: VTTC0801U) ← 매수와 다름!
        "custtype": "P",
        "hashkey": get_hashkey(data)
    }

    res = requests.post(URL, headers=headers, data=json.dumps(data))
    result = res.json()

    if result['rt_cd'] == '0':
        print(f"매도 주문 성공! 주문번호: {result['output']['ODNO']}")
    else:
        print(f"매도 주문 실패: {result['msg1']}")
    return result

def query_order_state():
    """매도 주문 - 매수와 거의 동일하되 tr_id만 다름"""
    PATH = "	/uapi/domestic-stock/v1/trading/inquire-psbl-rvsecncl"
    URL = f"{URL_BASE}/{PATH}"

    data = {
        "CANO": CANO,
        "ACNT_PRDT_CD": ACNT_PRDT_CD,
        "CTX_AREA_FK100": "",
        "CTX_AREA_NK100": "", 
        "INQR_DVSN_1": "0",
        "INQR_DVSN_2": "0"
    }

    headers = {
        "Content-Type": "application/json",
        "authorization": f"Bearer {ACCESS_TOKEN}",
        "appKey": APP_KEY,
        "appSecret": APP_SECRET,
        "tr_id": "TTTC0084R",             # 실전 매도 (모의: VTTC0801U) ← 매수와 다름!
        "custtype": "P"
    }

    res = requests.get(URL, headers=headers, data=json.dumps(data))
    result = res.json()

    if result['rt_cd'] == '0':
        print(f"조회 성공")
    else:
        print(f"조회 실패: {result['msg1']}")
    return result
```

```python
def order_rvsecncl(
    orgn_odno,              # 원주문번호(취소/정정 대상)
    ord_orgno="06010",      # 한국거래소전송주문조직번호(=주문점)
    mode="cancel",          # "cancel" or "revise"
    ord_dvsn="00",          # 00: 지정가, 01: 시장가 (원주문 성격과 맞추는 게 안전)
    ord_qty=0,
    ord_unpr=0,
    qty_all_ord_yn="Y",
    is_mock=False,
    get_hashkey=None,       # 해시키 함수가 있으면 넣고, 없으면 None
):
    url = f"{URL_BASE}/uapi/domestic-stock/v1/trading/order-rvsecncl"

    rvse_cncl = "02" if mode == "cancel" else "01"   # 취소=02, 정정=01  
    tr_id = "VTTC0803U" if is_mock else "TTTC0803U"  # 모의=VTTC0803U, 실전=TTTC0803U

    data = {
        "CANO": CANO,
        "ACNT_PRDT_CD": ACNT_PRDT_CD,
        "KRX_FWDG_ORD_ORGNO": ord_orgno,
        "ORGN_ODNO": str(orgn_odno),
        "ORD_DVSN": ord_dvsn,
        "RVSE_CNCL_DVSN_CD": rvse_cncl,
        "ORD_QTY": str(ord_qty),
        "ORD_UNPR": str(ord_unpr),
        "QTY_ALL_ORD_YN": qty_all_ord_yn,
    }  # 파라미터 구성 예시  :contentReference[oaicite:7]{index=7}

    headers = {
        "Content-Type": "application/json",
        "authorization": f"Bearer {ACCESS_TOKEN}",
        "appKey": APP_KEY,
        "appSecret": APP_SECRET,
        "tr_id": tr_id,
        "custtype": "P",
    }

    # 해시키가 필요한 환경이면 추가 (주문계열에서 요구되는 경우가 많음)
    if callable(get_hashkey):
        headers["hashkey"] = get_hashkey(data)

    r = requests.post(url, headers=headers, data=json.dumps(data), timeout=10)
    # 실패 원인 보려고 text 먼저 확보
    try:
        result = r.json()
    except Exception:
        raise RuntimeError(f"Non-JSON response: {r.status_code}\n{r.text}")

    return result
```
