```python
from pykrx import stock
import time
import pandas as pd

# 1) 날짜 범위 (예: 오늘 기준 2년)
end = pd.Timestamp.today().strftime("%Y%m%d")
start = (pd.Timestamp.today() - pd.DateOffset(years=2)).strftime("%Y%m%d")

days = stock.get_previous_business_days(fromdate=start, todate=end)
kospi = stock.get_market_ohlcv_by_ticker(days[0], market="KOSPI")
kospi
```
